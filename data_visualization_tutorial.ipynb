# Data Visualization Basics: Matplotlib and Seaborn

This notebook builds five chart types with both libraries using Seaborn's Tips dataset. The emphasis is on chart syntax and styling, not deep analysis.

## 1. Load and inspect the data

```python
import seaborn as sns
import matplotlib.pyplot as plt
from IPython.display import display, Image
from visualize_tips import create_matplotlib_charts, create_seaborn_charts

tips = sns.load_dataset("tips")
print(f"Rows: {len(tips)} | Columns: {len(tips.columns)}")
display(tips.head())
```

## 2. Build the charts

Each helper returns the saved PNG paths. Matplotlib gives fine-grained control over each chart element; Seaborn provides concise statistical plotting functions and easy grouping/palettes.

```python
matplotlib_files = create_matplotlib_charts(tips)
seaborn_files = create_seaborn_charts(tips)
print(f"Created {len(matplotlib_files)} Matplotlib charts and {len(seaborn_files)} Seaborn charts.")
```

## 3. Compare the Matplotlib versions

```python
for chart_path in matplotlib_files:
    print(chart_path.name)
    display(Image(filename=str(chart_path)))
```

## 4. Compare the Seaborn versions

```python
for chart_path in seaborn_files:
    print(chart_path.name)
    display(Image(filename=str(chart_path)))
```

## Takeaways

- Both libraries can create each chart type and save it as a PNG.
- Matplotlib exposes the axes and styling details directly.
- Seaborn uses DataFrame columns and semantic mappings (such as `hue`) to create statistical charts with less setup.
- The heatmaps use numeric columns only; category columns are intentionally excluded from correlation calculations.
